```python
# 20-07-2026 --- NLP Based Project 

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import nltk # NLTK - Natural Language Toolkit 
```

```python
# import the text file 
data = pd.read_csv('train.txt',sep=";", header=None ,names=['Statement', "Emotions"])
print(data)
```

```python
# basis functions to find the details of the given dataset 
# 1) head 
# print(data.head())

# 2) tail 
# print(data.tail())

# 3) sample 
# print(data.sample())

# isnull and info 
# print(data.info())

# isnull
# print(data.isnull().sum())

# describe
# print(data.describe()) # we have to learn about the frequency in strings 

# unique 
uniqueEmotions = data['Emotions'].unique()
print(uniqueEmotions)
```

```python
# Vectorization - to convert the text into number 
# we will create an empty dictionary so that we can assign the number to each emotions
emotion_number = {}
# now we will iterate each emotions from the array and add them to the dictionary after assigning unqiue number to each emotion
i = 0 # iterator 
for value in uniqueEmotions: 
    # print(value)
    emotion_number[value] = i 
    i +=1 

data['Emotions'] = data['Emotions'].map(emotion_number)
print(data)
print(emotion_number)
```

```python
# Data Cleaning 

# convert the text into lowercase 
# now we will use the lambda function to convert the letter into lowercasing 
data['Statement'] = data['Statement'].apply(lambda val: val.lower())
print(data['Statement'])
```

```python
# remove punctuations from the text 

import string 
# create a custom function to remove the punctuations

def remove_punctuations(text): 
    return text.translate(str.maketrans('', '', string.punctuation))

# now we will use the function to remove the punctuations 
data['Statement'] = data['Statement'].apply(remove_punctuations)
print(data['Statement'])
```

```python
# want to remove the numbers from the file 
def remove_numbers(txt): 
    new = ""
    for i in txt: 
        if not i.isdigit():  # isdigit() -- it will check the text is number or not 
            new = new + i
    return new 

data['Statement'] = data['Statement'].apply(remove_numbers)
print(data["Statement"])
```

```python
# we will write a code to remove the emojis from the text file 

def remove_emojis(txt): 
    new = ""
    for i in txt: 
        if i.isascii(): 
            new +=i 
    return new 

data["Statement"] = data["Statement"].apply(remove_emojis)
print(data["Statement"])
```

```python
# remove stopwords --- ex: is, the, was, and, in, etc...
# NLTK - Natural Language Toolkit 

# if we want to remove stopwords then we have to use NLTK which is used to perform the removing stop words like 'is', 'the', 'are', etc. 

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt') # punkt
nltk.download('stopwords') # stopwords
```

```python
# we want to see the group of words as stopwords which is present inside the english file
stop_words = set(stopwords.words('english'))
print(stop_words)
print(len(stop_words))
```

```python
# want to create a function so that we can remove the stopwords from the given sentence

def remove_stopwords(txt):  # txt act as a parameter 
    words = txt.split()  # it will use to split the text 
    cleaned = [] # here we will add those words which is not considered as a stopwords
    for i in words: 
        if not i in stop_words: 
            cleaned.append(i)
    return ' '.join(cleaned)
data["Statement"] = data["Statement"].apply(remove_stopwords)
```

```python
print(data["Statement"])
```

```python
# we will split the data into two parts (1) Training (2) Testing 
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data['Statement'], data['Emotions'], test_size=0.20, random_state=42)
```

```python
# Training Phase 
# X_train and y_train
print(f" X_train value is - {X_train} and y_train value is - {y_train}")
# print(y_train)

# Testing Phase 
print(f" X_test value is - {X_test} and y_test value is - {y_test}")
# print(y_test)
```

```python
# We will write the code for the bag of words and the TFIDF 

# Bag of Words 
from sklearn.feature_extraction.text import CountVectorizer 
vectorizer = CountVectorizer()
X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)
```

```python
# --- 5-08-2026 --- now we will learn about the Naive byes algorithm 

from sklearn.naive_bayes import MultinomialNB # import MultinomialNB 
from sklearn.metrics import accuracy_score # import accuracy_score so that we can find the accuracy of the metrics 

# create an object of the MultinomialNB 
nb_model = MultinomialNB()
```

```python
# now we will pass the X_train and y_train to the MultinomialNB object 
nb_model.fit(X_train_bow, y_train)  # here we are creating a multinomialNB for the given model 
```

```python
# now we will check the prediction for the countVectorizer in Bag of Words 
prediction_bow = nb_model.predict(X_test_bow)

# accuracy_score() function will be used for prediction 
print(accuracy_score(y_test, prediction_bow)) # so here the predicted value is 0.768125 - it means 76%
```
